# Titanic — Exploratory Data Analysis
Loads `data/train.csv` and produces charts saved to `plots/`.

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid', palette='muted')
PLOTS_DIR = os.path.join('..', 'plots')
os.makedirs(PLOTS_DIR, exist_ok=True)

df = pd.read_csv(os.path.join('..', 'data', 'train.csv'))
print('Shape:', df.shape)
df.head()
```

```python
print('--- dtypes ---')
print(df.dtypes)
print('\n--- missing values ---')
missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)
pd.DataFrame({'count': missing, 'pct': missing_pct}).loc[missing > 0]
```

```python
# Survival rate by Sex
fig, ax = plt.subplots(figsize=(6, 4))
df.groupby('Sex')['Survived'].mean().plot(kind='bar', ax=ax, color=['steelblue', 'salmon'])
ax.set_title('Survival Rate by Sex')
ax.set_ylabel('Survival Rate')
ax.set_xticklabels(ax.get_xticklabels(), rotation=0)
plt.tight_layout()
plt.savefig(os.path.join(PLOTS_DIR, 'survival_by_sex.png'), dpi=150)
plt.show()
```

```python
# Survival rate by Pclass
fig, ax = plt.subplots(figsize=(6, 4))
df.groupby('Pclass')['Survived'].mean().plot(kind='bar', ax=ax)
ax.set_title('Survival Rate by Pclass')
ax.set_ylabel('Survival Rate')
ax.set_xticklabels(['1st', '2nd', '3rd'], rotation=0)
plt.tight_layout()
plt.savefig(os.path.join(PLOTS_DIR, 'survival_by_pclass.png'), dpi=150)
plt.show()
```

```python
# Age distribution by survival
fig, ax = plt.subplots(figsize=(8, 4))
df[df['Survived'] == 0]['Age'].dropna().plot(kind='hist', bins=30, alpha=0.6, label='Not Survived', ax=ax)
df[df['Survived'] == 1]['Age'].dropna().plot(kind='hist', bins=30, alpha=0.6, label='Survived', ax=ax)
ax.set_title('Age Distribution by Survival')
ax.set_xlabel('Age')
ax.legend()
plt.tight_layout()
plt.savefig(os.path.join(PLOTS_DIR, 'age_distribution.png'), dpi=150)
plt.show()
```

```python
# Survival rate by Embarked
fig, ax = plt.subplots(figsize=(6, 4))
df.groupby('Embarked')['Survived'].mean().plot(kind='bar', ax=ax)
ax.set_title('Survival Rate by Embarked')
ax.set_ylabel('Survival Rate')
ax.set_xticklabels(ax.get_xticklabels(), rotation=0)
plt.tight_layout()
plt.savefig(os.path.join(PLOTS_DIR, 'survival_by_embarked.png'), dpi=150)
plt.show()
```

```python
# Fare distribution by survival
fig, ax = plt.subplots(figsize=(8, 4))
df[df['Survived'] == 0]['Fare'].plot(kind='hist', bins=40, alpha=0.6, label='Not Survived', ax=ax)
df[df['Survived'] == 1]['Fare'].plot(kind='hist', bins=40, alpha=0.6, label='Survived', ax=ax)
ax.set_title('Fare Distribution by Survival')
ax.set_xlabel('Fare')
ax.set_xlim(0, 300)
ax.legend()
plt.tight_layout()
plt.savefig(os.path.join(PLOTS_DIR, 'fare_distribution.png'), dpi=150)
plt.show()
```

```python
# Correlation heatmap (numeric columns)
numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
fig, ax = plt.subplots(figsize=(10, 7))
sns.heatmap(df[numeric_cols].corr(), annot=True, fmt='.2f', cmap='coolwarm', ax=ax)
ax.set_title('Correlation Heatmap')
plt.tight_layout()
plt.savefig(os.path.join(PLOTS_DIR, 'correlation_heatmap.png'), dpi=150)
plt.show()
print('All plots saved to plots/')
```
